# Getting Started
## Virtual Environment and Installing Requirements

As is conventional, it is recommended that the requirements for running this code be installed in a virtual environment. To make a virtual environment that will be physically tied to this directory (and deleted if this directory is ever deleted), run the following command in your terminal:

```bash
python3 -m venv .venv
```

This will create the virtual environment ```venv``` that will live in the hidden directory ```.venv``` inside of ```vdos/```. To activate and install the requirements without screwing up system-wide dependencies:

```bash
source .venv/bin/activate
pip install -r requirements.txt
```

With this, everything should be ready to go.
